## Part VI – Iterating Arrays (Lesson 6)

**15. Iteration (2 pts)**

- (a) Iterate over `np.array([3, 6, 9])` and print **each element multiplied by 2**.
- (b) Given `grid = np.array([[1, 2, 3], [4, 5, 6]])`, use a **nested for loop** to print every element one by one, then print the **total** of all elements (computed inside the loop).

````python
arr = np.array([3, 6, 9])

for item in arr:
    print(item * 2)

grid = np.array([[1, 2, 3], [4, 5, 6]])

total = 0

for i in range(len(grid)):
    for j in range(len(grid[i])):
        print(grid[i][j])
        total += grid[i][j]

print("Sum:", total)
````

## Part XI – Filtering Arrays (Lesson 11)

**24. Filtering (2 pts)**

Given `arr = np.array([12, 45, 67, 23, 89, 50, 34])`:
- (a) Using a **for loop**, build a boolean list `filter_arr` that keeps only values **greater than 40**. Print `filter_arr` and the filtered array.
- (b) Using the **direct (vectorized)** method, return only the **even** elements. Print the boolean mask and the filtered array.

````python
arr = np.array([12, 45, 67, 23, 89, 50, 34])

filter_arr = []

for number in arr:
    filter_arr.append(number > 40)

print(filter_arr)
print(arr[filter_arr])

even_numbers = arr % 2 == 0

print(even_numbers)
print(arr[even_numbers])
````

## Part XII – Random (Lesson 12)

**25. Random Numbers, Distribution and Permutation (2 pts)**

Import `random` from NumPy, then:
- (a) Generate and print a random **integer** from 0 to 49.
- (b) Generate a **2-D array with 2 rows and 4 columns** where each value is `1, 2, or 3` with probabilities `0.2, 0.5, 0.3`. Print it.
- (c) Create `deck = np.array([1, 2, 3, 4, 5])`. Print a random **permutation** of `deck`, then print `deck` to show it was not changed.

*(Outputs will vary every run.)*

````python
from numpy import random

number = random.randint(0, 50)
print(number)

values = random.choice([1, 2, 3], size=(2, 4), p=[0.2, 0.5, 0.3])
print(values)

deck = np.array([1, 2, 3, 4, 5])

shuffled_deck = random.permutation(deck)

print(shuffled_deck)
print(deck)
````
